# TASK 2: MACHINE LEARNING MODEL FOR PRICE PREDICTION 
___

## Objective

- To build a robust regression-based machine learning model that accurately predicts house prices using property features.

### Separate Target & Features

```python
# Target variable
y = df['SalePrice']

# Features
X = df.drop('SalePrice', axis=1)
```

## Data Preprocessing
___
### **Handle Missing Values**

- We will handle numerical and categorical features differently.

```python
# Separate numerical and categorical columns
num_cols = X.select_dtypes(include=['int64', 'float64']).columns
cat_cols = X.select_dtypes(include=['object']).columns
```

## Numerical: Fill with Median
## Categorical: Fill with Mode

```python
# Fill numerical missing values
for col in num_cols:
    X[col].fillna(X[col].median(), inplace=True)

# Fill categorical missing values
for col in cat_cols:
    X[col].fillna(X[col].mode()[0], inplace=True)
```

## Encode Categorical Variables

- We’ll use One-Hot Encoding.

```python
X_encoded = pd.get_dummies(X, drop_first=True)
```

## Train-Test Split

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X_encoded, y, test_size=0.2, random_state=42
)
```

## Feature Engineering
- Create New Features

```python
# Total square footage
X_train['TotalSF'] = (
    df.loc[X_train.index, 'TotalBsmtSF'] +
    df.loc[X_train.index, '1stFlrSF'] +
    df.loc[X_train.index, '2ndFlrSF']
)

X_test['TotalSF'] = (
    df.loc[X_test.index, 'TotalBsmtSF'] +
    df.loc[X_test.index, '1stFlrSF'] +
    df.loc[X_test.index, '2ndFlrSF']
)
```

```python
# House age
X_train['HouseAge'] = 2025 - df.loc[X_train.index, 'YearBuilt']
X_test['HouseAge'] = 2025 - df.loc[X_test.index, 'YearBuilt']
```

### Model Evaluation Function

```python
from sklearn.metrics import r2_score, mean_squared_error, mean_absolute_error
import numpy as np

def evaluate_model(model, X_train, X_test, y_train, y_test):
    model.fit(X_train, y_train)
    preds = model.predict(X_test)
    
    r2 = r2_score(y_test, preds)
    rmse = np.sqrt(mean_squared_error(y_test, preds))
    mae = mean_absolute_error(y_test, preds)
    
    return r2, rmse, mae
```

### Baseline Model – Linear Regression

```python
from sklearn.linear_model import LinearRegression

lr = LinearRegression()
lr_r2, lr_rmse, lr_mae = evaluate_model(lr, X_train, X_test, y_train, y_test)
```

## Advanced Models
___ 
### Random Forest Regressor

```python
from sklearn.ensemble import RandomForestRegressor

rf = RandomForestRegressor(
    n_estimators=200,
    random_state=42
)

rf_r2, rf_rmse, rf_mae = evaluate_model(rf, X_train, X_test, y_train, y_test)
```

### Gradient Boosting Regressor

```python
from sklearn.ensemble import GradientBoostingRegressor

gbr = GradientBoostingRegressor(
    n_estimators=200,
    learning_rate=0.05,
    random_state=42
)

gbr_r2, gbr_rmse, gbr_mae = evaluate_model(gbr, X_train, X_test, y_train, y_test)
```

## Model Comparison Table

```python
model_results = pd.DataFrame({
    'Model': ['Linear Regression', 'Random Forest', 'Gradient Boosting'],
    'R2 Score': [lr_r2, rf_r2, gbr_r2],
    'RMSE': [lr_rmse, rf_rmse, gbr_rmse],
    'MAE': [lr_mae, rf_mae, gbr_mae]
})

model_results
```

## Observations
___
### Model Performance Observations

- Linear Regression provides a baseline but struggles with complex patterns
- Random Forest significantly improves prediction accuracy
- Gradient Boosting achieves the best balance between bias and variance
- Ensemble models outperform linear models due to non-linear relationships
